```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
cars = pd.read_csv('./cars.csv')
cars.head(3)
```

```python
rows,cols = cars.shape
print(f"Rows:  {rows}\nCols: {cols}")
```

```python
cars.describe()
```

```python
cars.describe(include='all')
```

```python
cars.info()
```

```python
cars.nunique()
```

```python
cars.isna().sum()
```

```python
cars['selling_price'].fillna(cars['selling_price'].mode()[0], inplace=True)
```

```python
cars.size
```

```python
print(cols * rows)
```

```python
grouped = cars.groupby('transmission').size()
grouped
```

```python
sns.boxplot(data=cars)
```

```python
corr = cars.corr(numeric_only=True)
sns.heatmap(data=corr, annot=True)
plt.title('Correlation heatmap')
```

```python
corr = corr[['selling_price']]
sns.heatmap(data=corr, annot=True)
plt.title('Correlation with Selling Price')
```

```python
plt.pie(grouped, labels=grouped.index, autopct='%1.1f%%')
plt.title('Distribution of Cars by Fuel Type')
plt.show()
```
